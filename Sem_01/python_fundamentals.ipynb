## 1. Variables

Variables store information that a program can use later.

````python
# Integer
age = 22

# Float
height = 1.75

# String
name = "Alice"

# Boolean
is_student = True

print(age)
print(height)
print(name)
print(is_student)
````

### Data types

````python
print(type(age))
print(type(height))
print(type(name))
print(type(is_student))
````

## 2. Basic Operations

````python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)
````

## 3. Lists

Python commonly uses **lists** to store multiple values in a single variable.

````python
numbers = [5, 8, 12, 20]

print(numbers)
````

### Accessing elements

````python
print("First element:", numbers[0])
print("Third element:", numbers[2])
print("Last element:", numbers[-1])
````

### Modifying elements

````python
numbers[1] = 100
print(numbers)
````

### Adding elements

````python
numbers.append(50)
print(numbers)
````

### Finding the length

````python
print("Number of elements:", len(numbers))
````

## 4. Conditional Statements

### Basic `if` statement

````python
age = 20

if age >= 18:
    print("Adult")
````

### `if...else`

````python
temperature = 15

if temperature > 20:
    print("Warm")
else:
    print("Cold")
````

### `if...elif...else`

````python
grade = 82

if grade >= 90:
    print("A")
elif grade >= 80:
    print("B")
elif grade >= 70:
    print("C")
else:
    print("Needs improvement")
````

### Comparison operators

| Operator | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

### Logical operators

````python
age = 25
student = True

print(age > 18 and student)
print(age > 18 or student)
print(not student)
````

## 5. Loops

### `for` loop

````python
for i in range(5):
    print(i)
````

### Looping through a list

````python
colors = ["red", "blue", "green"]

for color in colors:
    print(color)
````

### `while` loop

````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

### Using `break`

````python
for i in range(10):
    if i == 5:
        break

    print(i)
````

### Using `continue`

````python
for i in range(10):
    if i % 2 == 0:
        continue

    print(i)
````

## 6. Functions

### A function without parameters

````python
def greet():
    print("Hello!")


greet()
````

### A function with a parameter

````python
def greet_person(name):
    print("Hello,", name)


greet_person("Alice")
````

### Returning a value

````python
def square(number):
    return number ** 2


result = square(5)
print(result)
````

### Multiple parameters

````python
def add(a, b):
    return a + b


print(add(3, 7))
````
