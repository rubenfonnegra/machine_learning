## Splitting nodes in a decision tree?

A classification tree tries to create groups that are as pure as possible.

### Gini impurity

$$Gini = 1 - \sum_{i=1}^{k} p_i^2$$

### Entropy

$$Entropy = -\sum_{i=1}^{k} p_i \log_2(p_i)$$

Here, $p_i$ is the proportion of observations belonging to class $i$. Lower impurity means a more homogeneous node.

### How to read a node

- **Condition**: the question used to split the data.
- **gini**: node impurity.
- **samples**: training observations in the node.
- **value**: observations from each class.
- **class**: predicted class.

## Advantages and Limitations

### Advantages

- Easy to understand and visualize.
- Requires little data preparation.
- Handles nonlinear relationships.
- Works for classification and regression.
- Can model feature interactions.

### Limitations

- Can overfit easily.
- Small data changes may produce a different tree.
- Greedy splits do not guarantee the globally best tree.
- Deep trees may become difficult to interpret.
- Feature importance can be misleading when features are correlated.
